```python
import pandas as pd

def summarize_model(df):
    mut_per_site = df.groupby("Site_ID").size()
    fixed = df["fixed"] == 1
    return [len(df), len(mut_per_site), (mut_per_site > 1).sum(), fixed.sum(), fixed.mean() * 100]

finite = pd.read_csv("finite_mutations_by_site_100.csv")
infinite = pd.read_csv("variants_infinite_100.csv")

m = ["mutation events", "unique sites", "sites with multiple mutations", "#fixed in human", "fixed mutations (%)"]
comparison = pd.DataFrame({"Finite": summarize_model(finite), "Infinite": summarize_model(infinite)}, index=m)
comparison
```
